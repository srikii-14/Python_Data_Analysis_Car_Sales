### Data Cleaning
    Find and fix missing or wrong data

```python
car['Engine HP'] = car['Engine HP'].fillna(car['Engine HP'].mode()[0])  #replacing missing data with the most commonly repeated data of that column
car['Number of Doors'] = car['Number of Doors'].fillna(car['Number of Doors'].mean())
car['Engine Cylinders'] = car['Engine Cylinders'].fillna(car['Engine Cylinders'].mean())#replacing missing data with the average data of that column
car['Engine Fuel Type'] = car['Engine Fuel Type'].fillna('unknown')# filling the missing values with the word unknown
car['Market Category'] = car['Market Category'].fillna('unknown')# filling the missing values with the word unknown
```

### Comparing Car brands with their highest sales

```python
selected_models = ['BMW', 'Audi','Ferrari', 'Ford','Nissan','Suzuki','Toyota',"Rolls-Royce"]
selected_years=range(2000,2017)
models=CAR[CAR["Brand"].isin(selected_models)&CAR['Year'].isin(selected_years)]
plt.figure(figsize=(15,10))
sns.scatterplot(data=models,x='Year',y='Car Price',hue='Brand', palette="Set2")
plt.title('highest sales')
plt.show()
```

The scatter plot shows that Rolls-Royce and Ferrari consistently have the highest car prices across all years, clearly separating themselves from the other brands.

BMW and Audi fall into the upper-mid price range with a steady upward trend over time, indicating strong premium positioning. 

In contrast, Toyota, Nissan, Suzuki, and Ford remain in the lower to mid-price range, showing more affordable and stable pricing throughout the years. 

Overall, the chart highlights a clear price gap between luxury brands and mass-market brands, and also indicates that car prices generally increase over the years, especially for premium models.
